<h2>Defaulter Analysis</h2>
<h2>Extracting defaulter records where the loan repayment failed (repay_fail == 1).</h2>

```python
df['repay_fail'].sum()
```

```python
df_defaulter = df[df['repay_fail'] == 1]
df_defaulter.head()
```

<h2>Demographic & Loan Categorization</h2>
1. Home Ownership
Distribution of defaulters categorized by home ownership status.

```python
df_home_ownership = (
df_defaulter.groupby('home_ownership')
    .size()
    .reset_index(name='Number of Defaulters')
    .sort_values(by='Number of Defaulters', ascending=False)
)
df_home_ownership.reset_index(inplace = True)
df_home_ownership.rename(columns = {'home_ownership': 'Home Ownership'}, inplace = True)

sum = df_home_ownership["Number of Defaulters"].sum()
df_home_ownership["Percentage of Defaulters"] = df_home_ownership["Number of Defaulters"].map(lambda x :round(((x/sum)*100),2))
df_home_ownership
```

2. Loan Purpose
Distribution of defaulters categorized by the purpose of the loan.

```python
df_purpose = (
df_defaulter.groupby('purpose')
    .size()
    .reset_index(name='Number of Defaulters')
    .sort_values(by='Number of Defaulters', ascending=False)
)
df_purpose.reset_index(inplace = True)
df_purpose.rename(columns = {'purpose': 'Purpose'}, inplace = True)

sum = df_purpose["Number of Defaulters"].sum()
df_purpose["Percentage of Defaulters"] = df_purpose["Number of Defaulters"].map(lambda x :round(((x/sum)*100),2))
df_purpose
```

3. Verification Status
Distribution of defaulters based on income/source verification status.

```python
df_verification_status = (
df_defaulter.groupby('verification_status')
    .size()
    .reset_index(name='Number of Defaulters')
    .sort_values(by='Number of Defaulters', ascending=False)
)
df_verification_status.reset_index(inplace = True)
df_verification_status.rename(columns = {'verification_status': 'Verification Status'}, inplace = True)

sum = df_verification_status["Number of Defaulters"].sum()
df_verification_status["Percentage of Defaulters"] = df_verification_status["Number of Defaulters"].map(lambda x :round(((x/sum)*100),2))
df_verification_status
```

4. Employment Length
Distribution of defaulters categorized by employment length.

```python
df_emp_length = (
df_defaulter.groupby('emp_length')
    .size()
    .reset_index(name='Number of Defaulters')
    .sort_values(by='Number of Defaulters', ascending=False)
)
df_emp_length.reset_index(inplace = True)
df_emp_length.rename(columns = {'emp_length': 'Employment_Length'}, inplace = True)

sum = df_emp_length["Number of Defaulters"].sum()
df_emp_length["Percentage of Defaulters"] = df_emp_length["Number of Defaulters"].map(lambda x :round(((x/sum)*100),2))
df_emp_length
```

5. Categorical Risk Analysis: Default Rates (%)
Note: Analyzing raw counts of defaulters can be misleading if category sizes differ significantly. The true risk indicator is the Default Rate (Defaulters ÷ Total Borrowers per category).

```python
sns.set(style="whitegrid")
fig, ax = plt.subplots(figsize=(10, 8))
fig.suptitle('Socio-economic factors causing default - Home Ownership', fontsize=20, color='darkblue')

df_plot = df_home_ownership[df_home_ownership['Percentage of Defaulters'] > 0]
colors = sns.color_palette('mako')[3:3+len(df_plot)]

ax.pie(
    x=df_plot['Percentage of Defaulters'],
    labels=df_plot['Home Ownership'],
    autopct='%1.0f%%',
    startangle=140,
    colors=colors
)
plt.show()
```

```python
f, axes = plt.subplots(1, figsize = (15,9))
f.suptitle('Socio-economic factors causing default - Purpose', fontsize = 30, color = 'darkblue')

fig1 = sns.barplot(
    data=df_purpose,
    x= 'Purpose', 
    y ="Percentage of Defaulters", 
    hue="Purpose", 
    dodge=False,   
    palette = 'mako'
)
axes.set_xlabel("Purpose Type", fontsize = 15)
for p in fig1.patches:
    height = (p.get_height())
    fig1.annotate('{}%'.format(height),xy=(p.get_x() + (p.get_width() / 2), height), ha='center', va='bottom')
```

```python
f, axes = plt.subplots(1, figsize = (20,9))
f.suptitle('Socio-economic factors causing default - Employment Length', fontsize = 30, color = 'darkblue')

fig1 = sns.barplot(
    data=df_emp_length,
    x= 'Employment_Length', 
    y ="Percentage of Defaulters", 
    hue="Employment_Length", 
    dodge=False,   
    palette = 'mako'
)
axes.set_xlabel("Employment Length", fontsize = 15)
for p in fig1.patches:
    height = (p.get_height())
    fig1.annotate('{}%'.format(height),xy=(p.get_x() + (p.get_width() / 2), height), ha='center', va='bottom')
```

<h1>Machine Modeling Preparation</h1>
Here we are spliting the data into a test and a triand set fo modeling.

```python
!pip install scikit-learn
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

y = numeric_df['repay_fail']
X = numeric_df.drop('repay_fail', axis = 1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 0)
#df_clean = df.copy()
```

<h1>Random Forest </h1>
we are using random forest classifier(a machine learnning model), to make prediction and evaluate how well they perform.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

RF = RandomForestClassifier()
RF.fit(X_train, y_train)

y_predict = RF.predict(X_test)

print(classification_report(y_test, y_predict))

RF_SC = accuracy_score(y_predict,y_test)
print(f"{round(RF_SC*100,2)}% Accurate")
```

checking and preparing the data for model to handle an imbalanced data.

```python
print("Data types in training set:")
print(X_train.dtypes.value_counts())

object_cols = X_train.select_dtypes(include=['object']).columns
print("\nNon-numeric columns:", list(object_cols))
```

<h1>Threshold creation and handling imbalance</h1>
since we discovered that the data had some imbalance, we are now taking this method of threshold creation to handle the imbalance and then evaluate the model.

```python
!pip install imbalanced-learn
from imblearn.ensemble import BalancedRandomForestClassifier
from sklearn.metrics import classification_report

brf = BalancedRandomForestClassifier(random_state=42)
brf.fit(X_train, y_train)

y_pred_brf = brf.predict(X_test)
print(classification_report(y_test, y_pred_brf))

y_probs = brf.predict_proba(X_test)[:, 1]

threshold = 0.35
y_pred_custom = (y_probs >= threshold).astype(int)

print(classification_report(y_test, y_pred_custom))
```

This is to handle the imbalance data.

```python
brf_tuned = BalancedRandomForestClassifier(
    n_estimators=200, 
    max_depth=12, 
    random_state=42
)
brf_tuned.fit(X_train, y_train)

y_pred_tuned = brf_tuned.predict(X_test)
print(classification_report(y_test, y_pred_tuned))
```
